```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("../datasets/raw/india_accidents_2022_2025.csv")

print("✅ Dataset Loaded")
print(df.shape)
```

```python
df.isnull().sum()
```

```python
print(df.duplicated().sum())
```

```python
df.drop_duplicates(inplace=True)

print(df.shape)
```

```python
df.columns = (
    df.columns
    .str.lower()
    .str.strip()
    .str.replace(" ", "_")
)

print(df.columns)
```

```python
df['date'] = pd.to_datetime(df['date'])

print(df['date'].head())
```

```python
df['month'] = df['date'].dt.month
df['year'] = df['date'].dt.year
df['day'] = df['date'].dt.day

print(df[['month', 'year', 'day']].head())
```

```python
df['is_night'] = (
    (df['hour'] >= 20) | (df['hour'] <= 5)
).astype(int)

print(df['is_night'].value_counts())
```

```python
df['high_risk'] = (
    df['accident_severity']
    .isin(['major', 'fatal'])
).astype(int)

print(df['high_risk'].value_counts())
```

```python
df.drop(
    columns=['accident_id', 'risk_score'],
    inplace=True
)

print(df.columns)
```

```python
df.to_csv(
    "../datasets/processed/master_accident_dataset.csv",
    index=False
)

print("✅ Master Dataset Saved")
```

```python
df['festival'] = df['festival'].fillna('No Festival')

print(df['festival'].value_counts())
```
